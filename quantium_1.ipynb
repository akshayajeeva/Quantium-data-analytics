```python
transactions['PACK_SIZE'] = (
    transactions['PROD_NAME']
    .str.extract(r'(\d+)')
    .astype(int)
)

transactions[['PROD_NAME', 'PACK_SIZE']].head()
```

```python
transactions['BRAND'] = transactions['PROD_NAME'].str.split().str[0]

transactions[['PROD_NAME', 'BRAND']].head()
```

```python
transactions['BRAND'].value_counts()
```

```python
data = pd.merge(
    transactions,
    customers,
    on='LYLTY_CARD_NBR',
    how='left'
)

data.head()
```

```python
print("Merged Data Shape:", data.shape)

data.info()
```

```python
sales_by_segment = data.groupby(
    ['LIFESTAGE', 'PREMIUM_CUSTOMER']
)['TOT_SALES'].sum().reset_index()

sales_by_segment.sort_values(by='TOT_SALES', ascending=False)
```

```python
customers_by_segment = data.groupby(
    ['LIFESTAGE', 'PREMIUM_CUSTOMER']
)['LYLTY_CARD_NBR'].nunique().reset_index(name='CUSTOMERS')

customers_by_segment
```

```python
transactions_by_segment = data.groupby(
    ['LIFESTAGE', 'PREMIUM_CUSTOMER']
)['TXN_ID'].count().reset_index(name='TRANSACTIONS')

transactions_by_segment
```

```python
avg_spend = data.groupby(
    ['LIFESTAGE', 'PREMIUM_CUSTOMER']
)['TOT_SALES'].mean().reset_index(name='AVG_SPEND')

avg_spend.sort_values(by='AVG_SPEND', ascending=False)
```

```python
avg_qty = data.groupby(
    ['LIFESTAGE', 'PREMIUM_CUSTOMER']
)['PROD_QTY'].mean().reset_index(name='AVG_QTY')

avg_qty
```

```python
data['PRICE_PER_PACK'] = data['TOT_SALES'] / data['PROD_QTY']

price_segment = data.groupby(
    ['LIFESTAGE', 'PREMIUM_CUSTOMER']
)['PRICE_PER_PACK'].mean().reset_index(name='AVG_PRICE')

price_segment.sort_values(by='AVG_PRICE', ascending=False)
```

```python
brand_sales = data.groupby('BRAND')['TOT_SALES'].sum().sort_values(ascending=False)

brand_sales.head(10)
```

```python
pack_sales = data.groupby('PACK_SIZE')['TOT_SALES'].sum().sort_values(ascending=False)

pack_sales
```
